```python
import os
import psutil
import humanize

# Mevcut işlem (Jupyter kernel) bilgilerini al
process = psutil.Process(os.getpid())

# Kullanılan bellek miktarını al (bayt cinsinden)
mem_info = process.memory_info().rss  # rss = resident set size

# Daha okunabilir hale getir
print(f"Aktif Jupyter Notebook kernel RAM kullanımı: {humanize.naturalsize(mem_info)}")
```

```python
import os
import psutil

process = psutil.Process(os.getpid())
mem_info = process.memory_info().rss  # bayt cinsinden

print(f"Aktif Jupyter Notebook kernel RAM kullanımı: {mem_info / (1024 ** 2):.2f} MB")
```
